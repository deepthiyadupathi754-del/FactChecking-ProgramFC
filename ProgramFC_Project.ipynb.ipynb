```python
%cd /content/ProgramFC

file_path = "models/program_execution.py"

with open(file_path, "r") as f:
    code = f.read()

# Fix CPU model loading
code = code.replace(
    "self.model.parallelize()",
    "self.model.to('cpu')"
)

# Fix Predict() parsing
code = code.replace(
    r"p1 = re.compile(r'Predict[(]\(.*?)[)]', re.S)",
    r"p1 = re.compile(r'Predict\((.*?)\)', re.S)"
)

with open(file_path, "w") as f:
    f.write(code)

print("Code compatibility updates applied.")

# Verify the changes
!grep -n "self.model.to" models/program_execution.py
!grep -n "Predict" models/program_execution.py
```

```python
%cd /content/ProgramFC

import json
import os

# Load the first HOVER development sample
with open("datasets/HOVER/claims/dev.json", "r") as f:
    data = json.load(f)

sample = data[0]

# Prepare a reasoning program for this claim
program = [
    'fact_1 = Verify("M.I.L.F.$ was recorded by Fergie and produced by Polow da Don.")',
    'fact_2 = Verify("M.I.L.F.$ was followed by Life Goes On.")',
    'label = Predict(fact_1 and fact_2)'
]

# Create the input format expected by ProgramFC
test_data = [{
    "idx": 0,
    "id": sample["id"],
    "claim": sample["claim"],
    "gold": sample["label"],
    "predicted_programs": [program]
}]

os.makedirs("demo_programs", exist_ok=True)

with open("demo_programs/hover_test.json", "w") as f:
    json.dump(test_data, f, indent=4)

print("Test program created successfully!")
print("Claim:", sample["claim"])
print("Actual label:", sample["label"])
print("Number of reasoning programs:", len(test_data[0]["predicted_programs"]))
```
